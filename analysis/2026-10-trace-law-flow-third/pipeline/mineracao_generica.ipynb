**G1-G6** descrevem uma unidade que passou na triagem visível da rodada atual. A parte genérica não escreve nem aprova a lição. Os CSVs legados e o balde silencioso não medido não são usados como entrada.

Defina `UNIDADE` no ambiente com uma candidata observada desta fonte. A fonte precisa passar intake; erro de leitura interrompe. Os outputs existentes nunca comprovam uma nova execução.

Campos ausentes ficam desconhecidos: ferramenta é a primeira chamada registrada, não necessariamente origem do valor; versão de agente não é versão do prompt; finalização observada não é sucesso. Evidências e localizadores de casos ficam privados.

## G0 · Carga e elegibilidade

A fachada observada reaplica as regras elegíveis à fonte, sem carregar tabela antiga. A unidade deve ter recorrência demonstrada nesta rodada. Modelo/prompt/chamada e `idx` não são reconstruídos por conveniência.

```python
import os
from base_pipeline import carregar_mineracao_observada
from mineracao_observada import perfil_unidade, gravar

UNIDADE = os.environ.get('UNIDADE')
assert UNIDADE, 'Defina UNIDADE com uma candidata da triagem observada desta fonte.'
A = carregar_mineracao_observada()
P = perfil_unidade(A, UNIDADE)
print('Unidade:', UNIDADE, '· rodada:', A.manifesto['rodada_id'][:16])
print('Somente visível observado; parte específica e gates ainda necessários.')
```

## G1 · Recorrência e escopo

Contagens de erros atribuídos, execuções, meses, papéis e ocorrências condicionais. O perfil usa o mesmo universo da triagem. Ausência de canal silencioso é não medição, não zero.

```python
display(P['perfil'])
assert len(P['casos']) == P['perfil']['erros_observaveis'].iloc[0]
print('Silenciosas não medidas; conteúdo da lição e destino ainda não validados.')
```

## G2 · Sub-unidades

Mecanismo, assinatura, papel e primeira ferramenta registrada no step. Não sabemos o inventário completo das chamadas nem necessariamente a ferramenta de origem do valor. NULL não vira nenhuma ferramenta. Estas divisões orientam investigação, não escrevem novas unidades automaticamente.

```python
display(P['sub_unidades'])
```

## G3 · Antes do erro

Usar o contexto exportado e vínculos compatíveis, nunca `idx - 1` nem enumeração do filtro. Predecessor ausente/não identificado fica desconhecido; fronteira de chamada não observada não vira primeira chamada.

```python
display(P['antes'])
print('Continuidade condicionada à extração; mesma chamada/namespace não demonstrados.')
```

## G4 · Depois do erro

Classificar somente o slot identificado: erro estruturado, suspeita, sem sinal observado ou desconhecido. Não usar `idx + 1`; ausência não prova último step. Flag final do próximo step é sinal operacional, não recuperação, qualidade da resposta ou prova de mesma chamada. Não atribuir baixa necessidade de memória apenas porque o vizinho ficou sem sinal.

```python
display(P['depois'])
display(P['casos']['proximo_final_observado'].value_counts(dropna=False).rename('erros').to_frame())
print('Não mede recuperação semântica, resposta correta ou orçamento da execução.')
```

## G5 · Estabilidade observada

Mês real e versão do agente estão disponíveis. Modelo e variante do prompt permanecem desconhecidos. Distribuição entre erros não demonstra aumento de taxa sem denominadores de exposição.

```python
display(P['estabilidade'])
print('Modelo e versão do prompt não exportados; não preencher com versão de agente.')
```

## G6 · Casos privados e preparação da investigação

População da candidata, não censo de ações. Localizadores preservam fonte, execução/agente/papel, `step_pos` e `step_ref`, sem inventar `idx`. A amostra determinística intercala papéis, até 30 casos; não é probabilística.

Nenhum caso ou identificador é exibido. A leitura aberta/fechada, concordância, auditoria independente e decisão do pesquisador são passos posteriores. Esta etapa não gera memória aprovada.

```python
SAIDA_RODADA = gravar(A, perfis=[UNIDADE])
print('Casos privados:', len(P['casos']), '· amostra:', len(P['amostra_investigacao']))
print('Rodada:', SAIDA_RODADA.name, '· sem exibir identificadores ou texto de casos')
print('Próximo gate: leitura independente e suficiência da evidência para redigir a proposta.')
```
